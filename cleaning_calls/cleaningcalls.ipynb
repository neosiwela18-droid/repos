```python
import pandas as pd
```

```python
df = pd.read_excel(r'C:\Users\acer\Documents\MEGA\Programming\ASPIRE\Python\data projects\simple_data_analy_viz\cleaning_calls\Customer Call List.xlsx', index_col=0)
```

```python
df
```

```python
df = df.drop_duplicates()
df
```

```python
df['Last_Name'] = df['Last_Name'].str.strip('123./_')
df = df.dropna(subset=['Last_Name'])
df
```

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('[^a-zA-Z0-9]', '', regex=True)
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:])
df['Phone_Number'] = df['Phone_Number'].str.replace('nan--', '')
df['Phone_Number'] = df['Phone_Number'].str.replace('Na--', '')
df
```

```python
df[['Street Adress', 'State', 'Zip']] = df['Address'].str.split(',' ,expand=True)
df
```

```python
df['Do_Not_Contact']= df['Do_Not_Contact'].replace('Y', 'Yes').replace('N', 'No')
df['Paying Customer']= df['Paying Customer'].replace('Y', 'Yes').replace('N', 'No')
df
```

```python
df = df.replace('N/a', '')
df = df.fillna('')
df
```

```python
df = df.drop(df[df['Do_Not_Contact'] == 'Yes'].index)
# df = df.drop(df[df['Phone_Number'] == ''].index)
# df = df.drop(df[df['Do_Not_Contact'] == 'Yes'].index)
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('', 'No')
df
```

```python
df = df.drop(df[df['Phone_Number'] == ''].index)
df
```
